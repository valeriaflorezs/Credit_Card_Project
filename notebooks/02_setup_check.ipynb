# 02. Setup Check



Comprueba que el proyecto esté bien configurado antes de correr los experimentos. La configuración usa semilla aleatoria 42 (para que los resultados se puedan reproducir) y validación cruzada anidada con 3 pliegues externos y 3 internos, y la variable objetivo es `default_payment_next_month`. Se probó la separación de variables y objetivo con un conjunto sintético de 5,000 filas (22.46 % de clase positiva) y se confirmó que el objetivo es binario. Finalmente, se verificó que el dataset real procesado (29,965 filas y 21 columnas) está disponible y contiene el objetivo y las columnas categóricas definidas en la configuración.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import numpy as np
import pandas as pd

from src import config
from src.synthetic_data import generate_synthetic_dataset
from src.data_loader import split_features_target

print('RANDOM_STATE:', config.RANDOM_STATE)
print('TARGET_COLUMN:', config.TARGET_COLUMN)
print('N_OUTER_FOLDS / N_INNER_FOLDS:', config.N_OUTER_FOLDS, '/', config.N_INNER_FOLDS)
```

```python
np.random.seed(config.RANDOM_STATE)

df = generate_synthetic_dataset()
print(df.shape)
print(df[config.TARGET_COLUMN].value_counts(normalize=True))
df.head()
```

```python
X, y = split_features_target(df)
print('X shape:', X.shape)
print('y shape:', y.shape)
assert set(y.unique()) == {0, 1}, 'El target debe ser binario'
print('OK: target binario, X/y separados correctamente.')
```

```python
from src.data_loader import load_processed_dataset

df_real = load_processed_dataset()
print('Dataset real encontrado:', df_real.shape)
assert config.TARGET_COLUMN in df_real.columns
assert set(config.CATEGORICAL_COLUMNS) <= set(df_real.columns)
print('OK: target y columnas categóricas de config.py presentes en el dataset real.')
```
