# EDA: Análisis Exploratorio de Datos

Este conjunto de datos Default of Credit Card Clients (UCI Machine Learning Repository) contiene información de 30,000 clientes de tarjetas de crédito en Taiwán, recolectada en 2005. El objetivo original del dataset es predecir si un cliente incurrirá en un incumplimiento de pago (default) al mes siguiente, a partir de su límite de crédito, información demográfica, historial de pagos de los últimos 6 meses y montos facturados en ese mismo periodo.

La variable objetivo es `default payment next month`, que indica si el cliente incurrió **(1)** o no **(0)** en incumplimiento de pago al mes siguiente. Esto permite plantear el problema como una tarea de **clasificación binaria**.

Este EDA revisa explícitamente los duplicados (y los **elimina**), trata las categorías no oficiales, controla los outliers antes de calcular correlación y VIF, y cada gráfico interactivo va acompañado de su interpretación, con una versión **básica** (palabras sencillas) y una **técnica**. En la Parte 4 se aplica y se guarda el preprocesamiento que resuelve todos los hallazgos, de modo que el dataset que reciben los modelos ya sale limpio de aquí.

`EDUCATION` trae tres códigos (**0, 5, 6**) que no forman parte de la documentación oficial (1=posgrado, 2=universidad, 3=secundaria, 4=otros); en conjunto representan **345 registros (1.2%)**. `MARRIAGE` trae un código **0** no documentado, con **54 registros (0.18%)**.

En ambos casos se trata de una fracción pequeña de la muestra, así que en vez de eliminarlas (lo que descartaría información del resto de variables de esos clientes) se **reagrupan dentro de la categoría "Otros"** ya existente (`EDUCATION = 4`, `MARRIAGE = 3`). Es una decisión de modelado, no está respaldada semánticamente: la fuente no dice que esos códigos signifiquen "otros", y sus tasas de default son distintas. En `EDUCATION`: código 0 (14 filas, 0% de default), 5 (280 filas, 6.4%), 6 (51 filas, 15.7%) y 4 documentado (123 filas, 5.7%). En `MARRIAGE`: código 0 (54 filas, 9.3%) frente a 3 (323 filas, 26.0%), que baja a 23.6% al fusionarlos.

Las variables `PAY_x` muestran una alta concentración en **0** (entre el 49% y el 57% de los casos según el mes) y en **-1** (18-20%), seguidos por **-2** (9-16%). Como se explicó arriba, la definición exacta de estos dos códigos no está estandarizada en la fuente original. Se conservan tal cual (no se recodifican ni se eliminan): el `0` representa a la mayoría de las observaciones y el `-2` a una minoría, y se desconoce si equivalen a clientes sin atraso. Se deja constancia de la ambigüedad para no sobre-interpretar resultados que dependan de la frontera exacta entre "0" y "-1".


```python
# Decisión de modelado (ver el texto anterior): estos códigos no documentados no son semánticamente "otros" y tienen tasas de default distintas.
df["EDUCATION"] = df["EDUCATION"].replace({0: 4, 5: 4, 6: 4})
df["MARRIAGE"] = df["MARRIAGE"].replace({0: 3})

display(df["EDUCATION"].value_counts().sort_index().rename("Clientes").rename_axis("EDUCATION (reagrupada)").to_frame())
df["MARRIAGE"].value_counts().sort_index().rename("Clientes").rename_axis("MARRIAGE (reagrupada)").to_frame()
```

## Análisis Univariado

### 1.1 Estadísticos descriptivos completos

Un valor $x$ se marca como atípico con el criterio de Tukey cuando cae fuera de las vallas

$$
x < Q_1 - 1.5\,\mathrm{IQR}
\quad \text{o} \quad
x > Q_3 + 1.5\,\mathrm{IQR},
\qquad \mathrm{IQR} = Q_3 - Q_1 .
$$

```python
def resumen_univariado(serie):
    s = serie.dropna()
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    lim_inf, lim_sup = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    outliers_pct = ((s < lim_inf) | (s > lim_sup)).mean() * 100
    return pd.Series({
        "n": s.shape[0],
        "media": s.mean(),
        "mediana": s.median(),
        "desv_estandar": s.std(),
        "min": s.min(),
        "Q1": q1,
        "Q3": q3,
        "max": s.max(),
        "IQR": iqr,
        "asimetria (skew)": stats.skew(s),
        "curtosis": stats.kurtosis(s),
        "% outliers (Tukey)": outliers_pct,
    })

tabla_univariada = pd.DataFrame({col: resumen_univariado(df[col]) for col in num_cols}).T
tabla_univariada.round(3)
```

### 1.2 Visualización interactiva

```python
# Las figuras guardan resúmenes (conteos por bin y cuartiles) en lugar de los 30,000 valores de cada
# variable: así el notebook pesa unos KB por figura y no varios MB, y el gráfico se ve igual.
# Bins y cuartiles se calculan con las mismas reglas que usa Plotly cuando recibe los datos crudos.

def bins_plotly(serie):
    """Bordes de los bins con la regla automática de Plotly (autoBin de plotly.js) para datos enteros:
    ancho «redondo» (1, 2 o 5 × 10^k) de al menos 2·σ/n^0.4, con bordes en medios enteros."""
    x = np.asarray(serie, dtype=float)
    lo, hi = x.min(), x.max()
    dif = np.diff(np.unique(x)).min()
    f = 10 ** np.floor(np.log10(dif))
    piso = f * max(v for v in (0.9, 1.9, 4.9, 9.9) if v <= dif / f + 1e-12)
    bruto = max(piso, 2 * x.std(ddof=1) / len(x) ** 0.4)
    base = 10 ** np.floor(np.log10(bruto))
    ancho = base * next(m for m in (1, 2, 5, 10) if m * base >= bruto - 1e-9 * base)
    inicio = np.floor(lo / ancho) * ancho
    if inicio == lo:
        inicio -= ancho
    if ancho < 1:
        inicio = lo - ancho / 2
    else:
        inicio -= 0.5
        if inicio + ancho < lo:
            inicio += ancho
    n = 1 + int(np.floor((hi - inicio) / ancho))
    return inicio + ancho * np.arange(n + 1)


def resumen_caja(serie):
    """Cuartiles como los calcula Plotly (sus cuartiles «linear» equivalen al método hazen de numpy) y
    bigotes hasta el mínimo y el máximo, que es lo que dibuja Plotly cuando boxpoints=False."""
    x = np.asarray(serie, dtype=float)
    q1, med, q3 = np.percentile(x, [25, 50, 75], method="hazen")
    return dict(q1=[q1], median=[med], q3=[q3], lowerfence=[x.min()], upperfence=[x.max()])


def fig_hist_box(dataframe, cols, height=560):
    n = len(cols)
    fig = make_subplots(rows=2, cols=1, shared_xaxes=True,
                         row_heights=[0.72, 0.28], vertical_spacing=0.04)
    for i, col in enumerate(cols):
        bordes = bins_plotly(dataframe[col])
        conteo, _ = np.histogram(dataframe[col], bordes)
        fig.add_trace(go.Bar(x=(bordes[:-1] + bordes[1:]) / 2, y=conteo, width=bordes[1] - bordes[0],
                             marker_color=DEEP_NAVY, marker_line_width=0, visible=(i == 0), showlegend=False,
                             customdata=np.column_stack([np.ceil(bordes[:-1]), np.floor(bordes[1:])]),
                             hovertemplate="%{customdata[0]:,.0f} a %{customdata[1]:,.0f}: %{y:,}<extra></extra>"),
                      row=1, col=1)
        fig.add_trace(go.Box(y=[""], orientation="h", name="", marker_color=WARM_GOLD,
                              visible=(i == 0), showlegend=False, **resumen_caja(dataframe[col])), row=2, col=1)
    botones = []
    for i, col in enumerate(cols):
        vis = [False] * (2 * n)
        vis[2 * i], vis[2 * i + 1] = True, True
        botones.append(dict(label=col, method="update",
                             args=[{"visible": vis}, {"title": f"Distribución de {col}"}]))
    fig.update_layout(
        title=f"Distribución de {cols[0]}", height=height, width=850, bargap=0,
        updatemenus=[dict(buttons=botones, direction="down", x=1.15, y=1.05, showactive=True)],
    )
    fig.update_xaxes(type="linear")   # con cajas precalculadas Plotly no deduce solo que el eje es numérico
    fig.update_xaxes(title_text="", row=1, col=1)
    fig.update_xaxes(title_text=cols[0], row=2, col=1)
    return fig
```

```python
# Interpretaciones de la sección 1.2 (variables numéricas). Los números se toman de
# `tabla_univariada` para que el texto siempre coincida con la tabla.
def nt(x):
    return f"-NT${abs(x):,.0f}" if x < 0 else f"NT${x:,.0f}"

PAY_COLS = ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]
BILL_COLS = [f"BILL_AMT{i}" for i in range(1, 7)]
PAYAMT_COLS = [f"PAY_AMT{i}" for i in range(1, 7)]
MESES_ANTES = {"PAY_0": 0, **{f"PAY_{k}": k - 1 for k in range(2, 7)},
               **{f"BILL_AMT{k}": k - 1 for k in range(1, 7)}, **{f"PAY_AMT{k}": k - 1 for k in range(1, 7)}}

def corresp(col):
    n = MESES_ANTES[col]
    return "al mes más reciente" if n == 0 else ("al mes anterior al más reciente" if n == 1 else f"a {n} meses antes del más reciente")

U = tabla_univariada
skew_max_global = U["asimetria (skew)"].idxmax()
out_pay = U.loc[PAY_COLS, "% outliers (Tukey)"]
skew_bill = U.loc[BILL_COLS, "asimetria (skew)"]
min_bill = U.loc[BILL_COLS, "min"]

def tecnica_uni(col):
    s = U.loc[col]
    sk, ku, out = s["asimetria (skew)"], s["curtosis"], s["% outliers (Tukey)"]
    if col == "LIMIT_BAL":
        return (f"Histograma con asimetría moderada a la derecha ({sk:.2f}); la mayoría de los clientes se concentra "
                f"entre {nt(s['Q1'])} y {nt(s['Q3'])} (rango intercuartílico), con una cola larga hasta {nt(s['max'])}; el boxplot "
                f"marca pocos outliers Tukey ({out:.2f}%) porque la asimetría es moderada y el rango intercuartílico es amplio respecto de la cola, no porque el criterio subestime el sesgo.")
    if col == "AGE":
        return (f"Distribución con asimetría moderada ({sk:.2f}, del mismo orden que la de `LIMIT_BAL`), concentrada entre {s['Q1']:.0f} y {s['Q3']:.0f} años "
                f"(mediana {s['mediana']:.0f}); cola derecha moderada hacia clientes mayores, con muy pocos outliers ({out:.1f}%).")
    if col in PAY_COLS:
        txt = (f"Variable discreta con la masa en 0 y -1 y una cola de atrasos de hasta {s['max']:.0f} meses; media {s['media']:.2f}, "
               f"mediana {s['mediana']:.0f}. El {out:.1f}% de los valores se marca como «outlier» bajo Tukey, un resultado artificial "
               f"de ser una variable discreta muy concentrada")
        if col == out_pay.idxmax():
            txt += " (la mayor proporción del bloque `PAY_x`)"
        elif col == out_pay.idxmin():
            txt += " (la menor proporción del bloque `PAY_x`)"
        txt += "."
        if col == "PAY_0":
            txt += " Es el mes más reciente del historial."
        else:
            prev = PAY_COLS[PAY_COLS.index(col) - 1]
            if s["media"] < U.loc[prev, "media"]:
                txt += f" La media es más negativa que la de `{prev}` ({U.loc[prev, 'media']:.2f}): cuanto más atrás en el tiempo, más clientes figuran al día. Esta tendencia está parcialmente confundida por la codificación inconsistente de `PAY_0` (3,667 clientes con valor 1, frente a 28, 4, 2, 0 y 0 en `PAY_2` a `PAY_6`)."
        return txt
    if col in BILL_COLS:
        txt = (f"Histograma extremadamente sesgado a la derecha (asimetría {sk:.2f}, curtosis {ku:.1f}); la mediana ({nt(s['mediana'])}) queda "
               f"muy por debajo de la media ({nt(s['media'])}). Hay valores negativos (mínimo {nt(s['min'])}) que reflejan saldos a favor del "
               f"cliente; {out:.1f}% de outliers Tukey.")
        if col == skew_bill.idxmax():
            txt += f" Es la más asimétrica del bloque por un valor extremo de {nt(s['max'])} que estira la cola derecha."
        if col == min_bill.idxmin():
            txt += " Incluye el valor negativo más extremo del dataset."
        return txt
    txt = (f"Distribución muy concentrada en montos bajos (mediana {nt(s['mediana'])} frente a una media de {nt(s['media'])}), con asimetría "
           f"{sk:.1f} y curtosis {ku:,.0f} por unos pocos pagos muy grandes (máximo {nt(s['max'])}) que distorsionan el promedio.")
    if col == skew_max_global:
        txt += " Es la variable con mayor asimetría de todo el dataset."
    return txt

def basica_uni(col):
    s = U.loc[col]
    if col == "LIMIT_BAL":
        return (f"Es el cupo de crédito que el banco le dio a cada cliente. La mayoría tiene entre {nt(s['Q1'])} y {nt(s['Q3'])}. "
                f"Pocos clientes tienen cupos muy altos (hasta {nt(s['max'])}), por eso el gráfico se estira hacia la derecha.")
    if col == "AGE":
        return (f"Casi todos los clientes tienen entre {s['Q1']:.0f} y {s['Q3']:.0f} años y la edad típica es {s['mediana']:.0f}. "
                f"A partir de ahí hay cada vez menos clientes a medida que aumenta la edad.")
    if col in PAY_COLS:
        base = ("Indica qué tan al día estaba el cliente con sus pagos: de 1 en adelante son meses de atraso, -1 es pago puntual, y -2 y 0 "
                "(que la fuente no define con claridad; su significado exacto es incierto) probablemente corresponden a clientes sin atraso, pero es una suposición. ")
        if col == "PAY_0":
            return base + "Corresponde al mes más reciente. La gran mayoría está al día; una minoría se atrasa y unos pocos llegan a 8 meses de atraso."
        return (f"Es el mismo dato que `PAY_0` (meses de atraso; -1 es pago puntual y -2 y 0 son ambiguos), pero correspondiente {corresp(col)}. "
                "Tiene la misma forma: casi todos al día y pocos con atrasos largos. Cuanto más atrás en el tiempo, más clientes aparecen al día, aunque en parte es un efecto de codificación: "
                "`PAY_0` tiene 3,667 clientes con valor 1, frente a 28, 4, 2, 0 y 0 en `PAY_2` a `PAY_6`.")
    if col in BILL_COLS:
        return (f"Es lo que el cliente debía según su estado de cuenta correspondiente {corresp(col)}. La mitad de los clientes debe menos de {nt(s['mediana'])}, "
                f"pero unos pocos deben cifras enormes, por eso el gráfico se estira hacia la derecha. Algunos valores son negativos: "
                f"son clientes que pagaron de más y tienen saldo a favor.")
    return (f"Es lo que el cliente pagó, correspondiente {corresp(col)}. La mitad de los clientes pagó menos de {nt(s['mediana'])}; "
            f"unos pocos pagaron cifras enormes y por eso el promedio ({nt(s['media'])}) engaña: es mucho más alto que lo habitual.")

INTERP_UNI_NUM = {c: {"basica": basica_uni(c), "tecnica": tecnica_uni(c)} for c in num_cols}
```

## Análisis Bivariado

### 2.1 Pruebas de hipótesis y comparación de grupos

Para cada variable numérica se compara el grupo sin *default* con el grupo con *default* mediante la U de Mann-Whitney, que no supone normalidad ($R_1$ es la suma de rangos del primer grupo y $n_1$ su tamaño), y la t de Welch, que admite varianzas distintas:

$$
U_1 = R_1 - \frac{n_1 (n_1 + 1)}{2},
\qquad
t = \frac{\bar{x}_1 - \bar{x}_2}{\sqrt{\dfrac{s_1^2}{n_1} + \dfrac{s_2^2}{n_2}}} .
$$

Para las variables categóricas se usa la prueba $\chi^2$ de independencia y la V de Cramér, con $k$ el menor número de categorías entre las dos variables:

$$
V = \sqrt{\frac{\chi^2 / n}{k - 1}} .
$$

```python
resultados_bivariado = []
for col in num_cols:
    g0 = df.loc[df[target] == 0, col]
    g1 = df.loc[df[target] == 1, col]
    u_stat, u_p = stats.mannwhitneyu(g0, g1, alternative="two-sided")
    t_stat, t_p = stats.ttest_ind(g0, g1, equal_var=False)
    resultados_bivariado.append({
        "variable": col,
        "media_no_default": g0.mean(), "media_default": g1.mean(),
        "mediana_no_default": g0.median(), "mediana_default": g1.median(),
        "Mann-Whitney U": u_stat, "p-value (MW)": u_p,
        "t (Welch)": t_stat, "p-value (t)": t_p,
    })

tabla_bivariada = pd.DataFrame(resultados_bivariado).set_index("variable")
tabla_bivariada.round(4)
```

```python
resultados_chi2 = []
for col in cat_cols:
    tabla_cruzada = pd.crosstab(df[col], df[target])
    chi2, p, dof, _ = stats.chi2_contingency(tabla_cruzada)
    resultados_chi2.append({"variable": col, "chi2": chi2, "p-value": p, "grados_libertad": dof})

tabla_chi2 = pd.DataFrame(resultados_chi2).set_index("variable")
tabla_chi2.round(6)
```

Con 30,000 observaciones, diferencias pequeñas resultan estadísticamente significativas: p cercano a 0 (Mann-Whitney p<0.001) solo en 14 de las 20 variables numéricas, y en las tres categóricas vía chi-cuadrado. No lo son `AGE` (p=0.40) ni `BILL_AMT2` a `BILL_AMT6` (p=0.006, 0.025, 0.137, 0.221 y 0.962). El p-value por sí solo aporta poco para priorizar variables, y estas pruebas no incluyen corrección por pruebas múltiples ni una medida de tamaño de efecto para las numéricas. En las categóricas, la V de Cramér es SEX 0.040, EDUCATION 0.073 y MARRIAGE 0.030: asociaciones muy débiles. Lo relevante es la **magnitud** de la diferencia entre medias/medianas y las tasas de default por categoría, que se exploran en los gráficos siguientes.

## Multicolinealidad

### 3.1 Matriz de correlación

Antes de calcular la correlación se **capan (winsorizan) los percentiles 1 y 99** de cada variable numérica. Esto evita que un puñado de valores extremos (p. ej. `BILL_AMT` o `LIMIT_BAL` en la cola superior) distorsione el coeficiente de correlación. La winsorización se aplica a todas las numéricas, incluidas las `PAY_x` ordinales (y, en la sección 4.2, `N_CREDIT_BALANCE_MONTHS`), y solo para correlación y VIF: el modelo no ve datos winsorizados.


La medida principal es la correlación de Spearman, que es la correlación de Pearson aplicada a los rangos y por eso mide asociación monótona, no exige linealidad ni normalidad y resiste valores extremos:

$$
\rho_s = \rho\!\left(\operatorname{rango}(X),\, \operatorname{rango}(Y)\right).
$$

Spearman es el único método de correlación del proyecto (el tablero y el resto del libro usan el mismo). Como Spearman solo depende del orden de los datos, la winsorización casi no la altera: los rangos de las correlaciones principales (por ejemplo, entre las `BILL_AMT`) coinciden con los de los datos originales.

```python
# Winsorización P1/P99 solo para correlación y VIF; incluye las PAY_x ordinales. El modelo no ve datos winsorizados.
def winsorize(serie, lower=0.01, upper=0.99):
    lo, hi = serie.quantile(lower), serie.quantile(upper)
    return serie.clip(lo, hi)

df_w = df[num_cols].apply(winsorize)

# Spearman es el único método de correlación del proyecto (mide asociación monótona con los rangos).
corr_spearman = df_w.corr(method="spearman")

fig = go.Figure(data=go.Heatmap(
    z=corr_spearman.values, x=corr_spearman.columns, y=corr_spearman.columns,
    colorscale=DIVERGING_NG, zmid=0, zmin=-1, zmax=1,
    text=corr_spearman.round(2).values, texttemplate="%{text}", textfont=dict(size=8),
    hovertemplate="%{y} vs %{x}<br>Spearman: %{z:.3f}<extra></extra>",
))
fig.update_layout(
    title="Matriz de correlación entre variables independientes (Spearman, winsorizada)",
    height=700, width=820,
)
fig.show()
```

Las seis variables de facturación (**`BILL_AMT1` a `BILL_AMT6`**) están **muy correlacionadas entre sí** (Spearman entre 0.73 y 0.91): tiene sentido, ya que el saldo de un mes está fuertemente relacionado con el saldo del mes siguiente si el cliente no paga el total. Las variables `PAY_0` a `PAY_6` también muestran correlación moderada-alta entre sí (miden el mismo fenómeno atraso en el pago en distintos meses). En cambio, los montos efectivamente pagados (`PAY_AMT*`) tienen una correlación moderada entre sí (Spearman entre 0.45 y 0.55) y más baja con las demás variables (hasta 0.32), salvo con las facturas de los meses cercanos (hasta 0.67). Esto ya anticipa un problema de multicolinealidad entre las variables de facturación, que se confirma con el VIF a continuación.


### 3.2 Factor de Inflación de la Varianza (VIF)

Para cada variable $j$, el VIF mide cuánto se infla la varianza de su coeficiente por la redundancia con las demás predictoras, donde $R_j^2$ es el coeficiente de determinación al regresar la variable $j$ sobre las otras:

$$
\mathrm{VIF}_j = \frac{1}{1 - R_j^2} .
$$

```python
# Datos winsorizados P1/P99 (también PAY_x ordinales), solo para el VIF; el modelo no los usa.
X_vif = df_w.copy()
X_vif = (X_vif - X_vif.mean()) / X_vif.std()
X_vif = X_vif.assign(const=1.0)

vif_data = pd.DataFrame({
    "variable": num_cols,
    "VIF": [variance_inflation_factor(X_vif.values, i) for i in range(len(num_cols))],
}).sort_values("VIF", ascending=False)

fig = px.bar(
    vif_data, x="VIF", y="variable", orientation="h",
    color="VIF", color_continuous_scale=[NAVY, PALE_GOLD, WARM_GOLD],
    title="Factor de Inflación de la Varianza (VIF) por variable (datos winsorizados)",
)
fig.add_vline(x=5, line_dash="dash", line_color=NAVY, annotation_text="VIF = 5")
fig.add_vline(x=10, line_dash="dash", line_color=WARM_GOLD, annotation_text="VIF = 10")
fig.update_layout(height=550, width=850, yaxis_title="", xaxis_title="VIF", coloraxis_showscale=False)
fig.show()

vif_data.round(2)
```

El VIF confirma lo que insinuaba la matriz de correlación: **`BILL_AMT2` a `BILL_AMT5` superan un VIF de 20**, muy por encima del umbral habitual de 10, y `BILL_AMT1`/`BILL_AMT6` rondan 13-15, evidencia fuerte de multicolinealidad severa entre las seis variables de facturación. Las variables `PAY_x` tienen un VIF moderado (3-4.5), por debajo del umbral de alerta pero mostrando cierta redundancia entre sí. El resto de variables (`LIMIT_BAL`, `AGE`, `PAY_AMT*`) tiene VIF cercano a 1-2, sin problema relevante.

Para un modelo lineal (regresión logística), conviene **no usar las seis `BILL_AMT` por separado**: alternativas razonables son quedarse solo con `BILL_AMT1` (el más reciente), construir un promedio o una pendiente/tendencia de los 6 meses, o aplicar una reducción de dimensionalidad (PCA) sobre ese bloque de variables. Para modelos basados en árboles (Random Forest, Gradient Boosting), la multicolinealidad afecta menos la performance predictiva, aunque sigue diluyendo la importancia atribuida a cada variable individual.

El dataset final tiene **29,965 filas y 21 columnas**: 20 variables predictoras y el target (`default_payment_next_month`, siempre la última columna).

## Conclusiones generales del EDA

- **Calidad de datos**: el dataset no tiene valores nulos. Las 35 copias excedentes se eliminan y las categorías no oficiales de `EDUCATION` y `MARRIAGE` se reagrupan en "Otros" (decisión de modelado, no respaldada semánticamente). - **Mejor variable predictora, por lejos**: el estado de pago del mes más reciente (`PAY_0`), seguido de los meses anteriores y del límite de crédito (`LIMIT_BAL`). Las variables demográficas (`AGE`, `SEX`, `EDUCATION`, `MARRIAGE`) aportan señal, pero mucho más débil.

- **Recomendación para el modelamiento**: dado que los atrasos (`PAY_x`) dominan la señal y su relación con el riesgo puede no ser lineal, un modelo basado en árboles (Random Forest, Gradient Boosting) es una opción natural. Si se prefiere un modelo lineal por interpretabilidad, conviene categorizar/discretizar `PAY_x`.
